```python
# EX 9: TEXT GENERATION USING LSTM NETWORKS

# Step 1: Import Libraries

import tensorflow as tf
import numpy as np


# Step 2: Load Shakespeare Dataset

path = tf.keras.utils.get_file(
    "shakespeare.txt",
    "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
)

text = open(
    path,
    "r",
    encoding="utf-8"
).read().lower()

# Use only a small portion for fast execution
text = text[:20000]

print("Characters used:", len(text))


# Step 3: Create Character Dictionary

chars = sorted(set(text))

c2i = {
    c: i for i, c in enumerate(chars)
}

i2c = {
    i: c for i, c in enumerate(chars)
}


# Step 4: Create Input and Output Sequences

seq_len = 30

X = []
y = []

# Use only 5000 sequences
for i in range(
    min(len(text) - seq_len, 5000)
):

    input_seq = text[
        i:i + seq_len
    ]

    target_char = text[
        i + seq_len
    ]

    X.append([
        c2i[c] for c in input_seq
    ])

    y.append(
        c2i[target_char]
    )


X = np.array(X)

y = np.array(y)

print("Training sequences:", len(X))


# Step 5: Build LSTM Model

model = tf.keras.Sequential([

    tf.keras.layers.Embedding(
        len(chars),
        32,
        input_length=seq_len
    ),

    tf.keras.layers.LSTM(
        64
    ),

    tf.keras.layers.Dense(
        len(chars),
        activation="softmax"
    )
])

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam"
)

model.summary()


# Step 6: Train the Model

print("\nTraining LSTM...")

model.fit(
    X,
    y,
    batch_size=128,
    epochs=1,
    verbose=1
)


# Step 7: Generate Text

def generate(seed, length=200):

    seed = seed.lower()

    # Make sure seed contains only known characters
    seed = "".join(
        c for c in seed
        if c in c2i
    )

    seq = [
        c2i[c]
        for c in seed
    ]

    generated = seed

    for _ in range(length):

        inp = np.array(
            seq[-seq_len:]
        ).reshape(1, -1)

        pred = model.predict(
            inp,
            verbose=0
        )[0]

        # Select next character
        next_idx = np.random.choice(
            len(pred),
            p=pred
        )

        seq.append(next_idx)

        generated += i2c[
            next_idx
        ]

    return generated


# Step 8: Generate and Display Text

print("\nGenerated Text:\n")

print(
    generate(
        "shall i compare thee to a summer's day?\n",
        200
    )
)
```
